```python
import requests
import json
```

```python
url = 'http://127.0.0.1:8001/chat'
query = 'add a comment to db "This pizza was the most delicious pizza I have ever eaten" with a rating 5 and username Hasret'

```

```python
params = {'message' : query,
            'verbose' : True}
%time response = requests.post(url, json = params)

print(response.json())
```

```python
query_fetch = 'fetch all the comemnts'
```

```python
params_fetch = {'message' : query_fetch,
            'verbose' : False}

%time response = requests.post(url, json = params_fetch)

print(response.json())
```
